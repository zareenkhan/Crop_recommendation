```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score

# Sample data (replace with actual data)
data =pd.read_csv("Crop_recommendation.csv")
# Create a DataFrame from the sample data
df = pd.DataFrame(data)

# Define the crop_dict (mapping labels to integers)
crop_dict = {
    'rice': 1, 'maize': 2, 'jute': 3, 'cotton': 4, 'coconut': 5, 'papaya': 6,
    'orange': 7, 'apple': 8, 'muskmelon': 9, 'watermelon': 10, 'grapes': 11,
    'mango': 12, 'banana': 13, 'pomegranate': 14, 'lentil': 15, 'blackgram': 16,
    'mungbean': 17, 'mothbeans': 18, 'pigeonpeas': 19, 'kidneybeans': 20, 'chickpea': 21, 'coffee': 22
}
# Add the crop_num column
df['crop_num'] = df['label'].map(crop_dict)
# Split the data into X (features) and Y (target)
X = df[['N', 'P', 'K', 'temperature', 'humidity', 'ph', 'rainfall']]
Y = df['crop_num']

# Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Split into training and test data
X_train, X_test, Y_train, Y_test = train_test_split(X_scaled, Y, test_size=0.2, random_state=42)

# Train a RandomForestClassifier
model = KNeighborsClassifier(n_neighbors=7)
model.fit(X_train, Y_train)

# Evaluate the model
Y_pred = model.predict(X_test)
accuracy = accuracy_score(Y_test, Y_pred)
print(f'Accuracy: {accuracy * 100:.2f}%')
```

```python
import gradio as gr

# Define the prediction function
def predict_crop(N, P, K, temperature, humidity, ph, rainfall):
    # Prepare the input features
    input_features = [[N, P, K, temperature, humidity, ph, rainfall]]

    # Scale the input
    input_scaled = scaler.transform(input_features)

    # Predict the crop number
    predicted_crop_num = model.predict(input_scaled)[0]

    # Reverse the crop_num to crop name using crop_dict
    predicted_crop = [k for k, v in crop_dict.items() if v == predicted_crop_num][0]

    return predicted_crop

# Create the Gradio interface using the new syntax
interface = gr.Interface(
    fn=predict_crop,
    inputs=[
        gr.Slider(minimum=0, maximum=200, step=1, label="Nitrogen"),
        gr.Slider(minimum=0, maximum=100, step=1, label="Phosphorus"),
        gr.Slider(minimum=0, maximum=100, step=1, label="Potassium"),
        gr.Slider(minimum=-10, maximum=50, step=0.1, label="Temperature (°C)"),
        gr.Slider(minimum=0, maximum=100, step=0.1, label="Humidity (%)"),
        gr.Slider(minimum=0, maximum=14, step=0.1, label="pH"),
        gr.Slider(minimum=0, maximum=300, step=1, label="Rainfall (mm)"),
    ],
    outputs="text",
    live=True
)

# Launch the interface
interface.launch()
```
